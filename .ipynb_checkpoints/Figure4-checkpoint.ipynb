### Imports and plotting setting

```python
import numpy as np
from scipy.optimize import fsolve, root

# import predefined functions wich solve the box model
import fct_BoxModel as fbm

import xarray as xr
import pandas as pd
import matplotlib
from matplotlib import cm
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from matplotlib.colors import Normalize
import cv2 as cv

from matplotlib import colors

from cmcrameri import cm as cmcra #berlin colormap

matplotlib.rcParams['mathtext.fontset'] = 'custom'
matplotlib.rcParams['mathtext.rm'] = 'Bitstream Vera Sans'
matplotlib.rcParams['mathtext.it'] = 'Bitstream Vera Sans:italic'
matplotlib.rcParams['mathtext.bf'] = 'Bitstream Vera Sans:bold'

matplotlib.rcParams['mathtext.fontset'] = 'stix'
matplotlib.rcParams['font.family'] = 'STIXGeneral'

SMALL_SIZE = 12
MEDIUM_SIZE = 14
BIGGER_SIZE = 16

plt.rc('font', size=SMALL_SIZE)          # controls default text sizes
plt.rc('axes', titlesize=SMALL_SIZE)     # fontsize of the axes title
plt.rc('axes', labelsize=MEDIUM_SIZE)    # fontsize of the x and y labels
plt.rc('xtick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('ytick', labelsize=SMALL_SIZE)    # fontsize of the tick labels
plt.rc('legend', fontsize=SMALL_SIZE)    # legend fontsize
plt.rc('figure', titlesize=BIGGER_SIZE)  # fontsize of the figure title

def truncate_colormap(cmap, minval=0.0, maxval=1.0, n=128):
    new_cmap = colors.LinearSegmentedColormap.from_list(
        'trunc({n},{a:.2f},{b:.2f})'.format(n=cmap.name, a=minval, b=maxval),
        cmap(np.linspace(minval, maxval, n)))
    return new_cmap

lipari_cut = truncate_colormap(cmcra.lipari, minval=0.15, maxval=1.0, n=128)
```

### Open usefull datasets

```python
inputs = xr.open_dataset("model_input_dataset.nc")
index_isf = inputs.Nisf.values
name_isf = inputs.name_isf.values
```

### Define Default parameters

```python
#PICO parameters
# Reese 2018
C_ref = 1e6 #m^6/s/kg
gammaT_ref = 2e-5 #m/s

#Vertical mixing
# ~Olbers & Hellmer 2010
kappa_diff_ref = 1e-7  # m/s
kappa_conv_ref = 1e-3 # m/s

#Polynia
g_ref = 20

# Continuous mixing
Lamb_ref = 0.3 #kg/m^3

#AABW
C2_ref = 4e6 #m^6/s/kg


#### Important line: choose your ice shelf (11=Filchner-Ronne, 66=Pine Island)
index = 66


nbox = int(inputs.Nbox.sel(Nisf=index).values)
Ac = inputs.Ac.sel(Nisf=index).values #m^2
frac = inputs.frac.sel(Nisf=index).values
depth = inputs.depth.sel(Nisf=index).values

AABW_ref = False # If you want to use the AABW param
T_surf_ref = np.mean(inputs.T_surf.sel(Nisf=index).values)
S_surf_ref = np.mean(inputs.S_surf.sel(Nisf=index).values)

Ap_ref = np.mean(inputs.Ap.sel(Nisf=index).values)
Omega_ref = np.mean(inputs.Omega.sel(Nisf=index).values)/30/24/3600  #m/s
T0_ref = np.mean(inputs.T0.sel(Nisf=index).values)
S0_ref = np.mean(inputs.S0.sel(Nisf=index).values)
```

### Define functions to make melt maps for the step function parameterization of $\kappa$

```python
def run_BoxModel(arguments, max_iter, guess, random_step):

    X, info, ier, msg = fsolve(fbm.BoxModel, guess, args = tuple(arguments), full_output=True,)
        
    if ier != 1: #management of non convergence cases
        incr = 0
        while ier != 1:
            X, info, ier, msg = fsolve(fbm.BoxModel, np.random.normal(guess,random_step), args = tuple(arguments), full_output=True)
            incr += 1
            if incr == max_iter:
                break

    return X, ier



def melt_map(Ap_list, param_list, param_position, max_iter, random_step, regime):

    guess_conv_0 = np.array([-1.9,T0_ref,S0_ref,S0_ref])
    guess_diff_0 = guess_conv_0
    
    stability_d = np.zeros((Npoint,Npoint))
    stability_c = np.zeros((Npoint,Npoint))
    
    melt_d = np.zeros((Npoint,Npoint))
    melt_c = np.zeros((Npoint,Npoint))
    
    sigma_d = np.zeros((Npoint,Npoint))
    sigma_c = np.zeros((Npoint,Npoint))
   
    for m, Ap in enumerate(Ap_list):

        guess_conv = guess_conv_0
        guess_diff = guess_diff_0

        for n,param in enumerate(param_list):

            # setting values of PICO parameters to ccompute melt rates
            if param_position == 1:
                C = param
            else:
                C = C_ref
            if param_position == 2:
                gammaT = param
            else:
                gammaT = gammaT_ref

            # manage the case where we want to play on kappa
            if param_position == 8:
                if regime == "diff":
                    kappa_diff = param
                    kappa_conv = kappa_conv_ref
                elif regime == "conv":
                    kappa_conv = param
                    kappa_diff = kappa_diff_ref
            else:
                kappa_diff = kappa_diff_ref
                kappa_conv = kappa_conv_ref
                
            ##################
            # Diffusive case #
            ##################
            arguments = [nbox,C_ref,gammaT_ref,Ac,Omega_ref,T0_ref,S0_ref,Ap,kappa_diff,g_ref,frac,depth, C2_ref, T_surf_ref, S_surf_ref, AABW_ref]
            if param_position != 8: #exclude kappa case
                arguments[param_position] = param
            
            X_d, ier = run_BoxModel(arguments, max_iter, guess_diff, random_step)

            # Updating initial guess
            if ier == 1:
                guess_diff = X_d
            else:
                print('No convergence Diff',m,n)
            if m==0:
                guess_diff_0 = X_d


            #Filling the tab
            if ier==1:
                sigma_d[m,n] = fbm.compute_Sigma(X_d)
                stability_d[m,n] = fbm.compute_Sigma(X_d)<0
                if stability_d[m,n]:
                    T,S,melt,q = fbm.PICO(X_d[1],X_d[3], nbox, C, gammaT, Ac, frac, depth)
                    melt_d[m,n] = fbm.compute_m_avg(melt,frac,nbox)
                else:
                    melt_d[m,n] = np.nan
            else:
                stability_d[m,n] = np.nan
                sigma_d[m,n] = np.nan

            ###################
            # Convective case #
            ###################
            arguments = [nbox,C_ref,gammaT_ref,Ac,Omega_ref,T0_ref,S0_ref,Ap,kappa_conv,g_ref,frac,depth, C2_ref, T_surf_ref, S_surf_ref, AABW_ref]
            if param_position != 8: #exclude kappa case
                arguments[param_position] = param

            X_c, ier = run_BoxModel(arguments, max_iter, guess_conv, random_step)
        
            # Updating initial guess
            if ier == 1:
                guess_conv = X_c
            else:
                print('No convergence Conv',m,n)
            if m==0:
                guess_conv_0 = X_c

            if ier==1:
                sigma_c[m,n] = fbm.compute_Sigma(X_c)
                stability_c[m,n] = fbm.compute_Sigma(X_c)>0
                if stability_c[m,n]:
                    T,S,melt,q = fbm.PICO(X_c[1],X_c[3], nbox, C_ref, gammaT_ref, Ac, frac, depth)
                    melt_c[m,n] = fbm.compute_m_avg(melt,frac,nbox)
                else:
                    melt_c[m,n] = np.nan
            else:
                stability_c[m,n] = np.nan
                sigma_c[m,n] = np.nan

    bistable = np.ma.masked_where(np.logical_not(np.logical_and(stability_c,stability_d)),np.logical_and(stability_c,stability_d)*0.25)

    return melt_d, melt_c, sigma_d, sigma_c, stability_d, stability_c, bistable
```

### Define functions to make melt maps for the smooth parameterization of $\kappa$

```python
def run_BoxModel_smooth(arguments, max_iter, guess, random_step):

    X, info, ier, msg = fsolve(fbm.BoxModel_continuous_kappa, guess, args = tuple(arguments), full_output=True,)
        
    if ier != 1: #management of non convergence cases
        incr = 0
        while ier != 1:
            X, info, ier, msg = fsolve(fbm.BoxModel_continuous_kappa, np.random.normal(guess,random_step), args = tuple(arguments), full_output=True)
            incr += 1
            if incr == max_iter:
                break

    return X, ier



def melt_map_smooth(Ap_list, param_list, param_position, max_iter, random_step, tol):

    guess_fw_0 = np.array([-1.9,T0_ref,S0_ref,S0_ref])
    guess_bw_0 = guess_fw_0
    
    melt_fw = np.zeros((Npoint,Npoint))
    melt_bw = np.zeros((Npoint,Npoint))
    
    sigma_fw = np.zeros((Npoint,Npoint))
    sigma_bw = np.zeros((Npoint,Npoint))
   
    
    for n,param in enumerate(param_list):

        # setting values of PICO parameters to ccompute melt rates
        if param_position == 1:
            C = param
        else:
            C = C_ref
        if param_position == 2:
            gammaT = param
        else:
            gammaT = gammaT_ref

        guess_fw = guess_fw_0
        guess_bw = guess_bw_0


        ##################
        #     Forward    #
        ##################
        for m, Ap in enumerate(Ap_list):
                
            arguments = [nbox,C_ref,gammaT_ref,Ac,Omega_ref,T0_ref,S0_ref,Ap, kappa_diff_ref, kappa_conv_ref, Lamb_ref,g_ref,frac,depth, C2_ref, T_surf_ref, S_surf_ref, AABW_ref]
            arguments[param_position] = param
            
            X, ier = run_BoxModel_smooth(arguments, max_iter, guess_fw, random_step)

            # Updating initial guess
            if ier == 1:
                guess_fw = X
                if m==0:
                    guess_fw_0 = X

                sigma_fw[m,n] = fbm.compute_Sigma(X)
                Tc,Sc,melt,q = fbm.PICO(X[1], X[3], nbox, C, gammaT, Ac, frac, depth)
                m_avg = fbm.compute_m_avg(melt,frac,nbox)
                melt_fw[m,n] = m_avg 
            
            else:
                print('No convergence FW',m,n)
                sigma_fw[m,n] = np.nan
                melt_fw[m,n] = np.nan


        ###################
        #     Backward    #
        ###################
        for m, Ap in enumerate(np.flip(Ap_list)):
            m = Npoint-m-1
                    
            arguments = [nbox,C_ref,gammaT_ref,Ac,Omega_ref,T0_ref,S0_ref,Ap, kappa_diff_ref, kappa_conv_ref, Lamb_ref,g_ref,frac,depth, C2_ref, T_surf_ref, S_surf_ref, AABW_ref]
            arguments[param_position] = param
            
            X, ier = run_BoxModel_smooth(arguments, max_iter, guess_bw, random_step)

            # Updating initial guess
            if ier == 1:
                guess_bw = X
                if m==Npoint-1:
                    guess_bw_0 = X

                sigma_bw[m,n] = fbm.compute_Sigma(X)
                Tc,Sc,melt,q = fbm.PICO(X[1], X[3], nbox, C, gammaT, Ac, frac, depth)
                m_avg = fbm.compute_m_avg(melt,frac,nbox)
                melt_bw[m,n] = m_avg 
            
            else:
                print('No convergence BW',m,n)
                sigma_bw[m,n] = np.nan
                melt_bw[m,n] = np.nan

    monostable = np.ma.masked_where(np.abs(sigma_fw-sigma_bw)>tol, melt_fw)
    bistable = np.ma.masked_where(np.abs(sigma_fw-sigma_bw)<tol, np.ones((Npoint,Npoint))*0.25)

    return monostable, bistable, sigma_fw, sigma_bw
```

```python
# The code use randomness to solve the non linear equations
np.random.seed(0) #this seed works well

fig, ax = plt.subplots(2,3,figsize=(9.8,6.8),sharey=True)

fig.suptitle('Pine Island')
vmax = 2

##################################
#      Ap x gammaT^star plot     #
##################################

Ap_min = 0
Ap_max = 8.5e9 #m^2
Npoint = 100
Ap_list = np.linspace(Ap_min,Ap_max,Npoint)

gammaT_list = np.logspace(np.log10(0.2),np.log10(20),Npoint)*1e-5
gammaT_position = 2

melt_d, melt_c, sigma_d, sigma_c, stability_d, stability_c, bistable = melt_map(Ap_list, gammaT_list, gammaT_position, max_iter=200, random_step=0.1,regime='')

plot1 = ax[0,0].pcolormesh(gammaT_list,Ap_list/1e9,melt_c,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[0,0].pcolormesh(gammaT_list,Ap_list/1e9,melt_d,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[0,0].pcolormesh(gammaT_list,Ap_list/1e9,bistable,cmap='binary',vmin=0,vmax=1) # Ap in km²


if np.max(sigma_d)>0:
    ax[0,0].contour(gammaT_list,Ap_list/1e9,sigma_d,colors='black',levels=0)
    ax[0,0].contourf(gammaT_list,Ap_list/1e9,np.ma.masked_where(np.logical_and(np.logical_not(stability_c),np.logical_not(stability_d)),sigma_d),hatches='/',levels=[0,500],alpha=0)
if np.min(sigma_c)<0:
    ax[0,0].contour(gammaT_list,Ap_list/1e9,sigma_c,colors='black',levels=0,linestyles='dashed')

ax[0,0].scatter(gammaT_ref,Ap_ref/1e9,facecolors='white',edgecolors='black',marker='d',s=50,linewidths=1)

ax[0,0].set_xscale('log')
ax[0,0].set_xlabel(r'$\gamma_T^\star$'+' [m/s]')
ax[0,0].set_ylabel(r'$\mathcal{A}_p$'+' '+r'$\mathrm{[\times 10^3 km^2]}$')


########################
#      Ap x C plot     #
########################

C_list = np.logspace(-1,np.log10(30),Npoint)*1e6
C_position = 1

melt_d, melt_c, sigma_d, sigma_c, stability_d, stability_c, bistable = melt_map(Ap_list, C_list, C_position, max_iter=200, random_step=0.1,regime='')

ax[0,1].pcolormesh(C_list,Ap_list/1e9,melt_c,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[0,1].pcolormesh(C_list,Ap_list/1e9,melt_d,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[0,1].pcolormesh(C_list,Ap_list/1e9,bistable,cmap='binary',vmin=0,vmax=1) # Ap in km²


if np.max(sigma_d)>0:
    ax[0,1].contour(C_list,Ap_list/1e9,sigma_d,colors='black',levels=0)
    ax[0,1].contourf(C_list,Ap_list/1e9,np.ma.masked_where(np.logical_and(np.logical_not(stability_c),np.logical_not(stability_d)),sigma_d),hatches='/',levels=[0,500],alpha=0)
if np.min(sigma_c)<0:
    ax[0,1].contour(C_list,Ap_list/1e9,sigma_c,colors='black',levels=0,linestyles='dashed')

ax[0,1].scatter(C_ref,Ap_ref/1e9,facecolors='white',edgecolors='black',marker='d',s=50,linewidths=1)

ax[0,1].set_xscale('log')
ax[0,1].set_xlabel(r'$C$'+' '+r'$\mathrm{[m^6/kg/s]}$')


########################
#      Ap x g plot     #
########################

g_list = np.logspace(0,3,Npoint)
g_position = 9

melt_d, melt_c, sigma_d, sigma_c, stability_d, stability_c, bistable = melt_map(Ap_list, g_list, g_position, max_iter=200, random_step=0.1,regime='')

ax[0,2].pcolormesh(g_list,Ap_list/1e9,melt_c,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[0,2].pcolormesh(g_list,Ap_list/1e9,melt_d,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[0,2].pcolormesh(g_list,Ap_list/1e9,bistable,cmap='binary',vmin=0,vmax=1) # Ap in km²


if np.max(sigma_d)>0:
    ax[0,2].contour(g_list,Ap_list/1e9,sigma_d,colors='black',levels=0)
    ax[0,2].contourf(g_list,Ap_list/1e9,np.ma.masked_where(np.logical_and(np.logical_not(stability_c),np.logical_not(stability_d)),sigma_d),hatches='/',levels=[0,500],alpha=0)
if np.min(sigma_c)<0:
    ax[0,2].contour(g_list,Ap_list/1e9,sigma_c,colors='black',levels=0,linestyles='dashed')

ax[0,2].scatter(g_ref,Ap_ref/1e9,facecolors='white',edgecolors='black',marker='d',s=50,linewidths=1)

ax[0,2].set_xscale('log')
ax[0,2].set_xlabel(r'$\tilde{\gamma}_T/\gamma_T^\star$')


################################
#     Ap x kappa_diff plot     #
################################

kappa_list = np.logspace(-8,-3,Npoint)
kappa_position = 8

melt_d, melt_c, sigma_d, sigma_c, stability_d, stability_c, bistable = melt_map(Ap_list, kappa_list, kappa_position, 200, 0.1, regime="diff")

ax[1,0].pcolormesh(kappa_list,Ap_list/1e9,melt_c,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[1,0].pcolormesh(kappa_list,Ap_list/1e9,melt_d,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[1,0].pcolormesh(kappa_list,Ap_list/1e9,bistable,cmap='binary',vmin=0,vmax=1) # Ap in km²


if np.max(sigma_d)>0:
    ax[1,0].contour(kappa_list,Ap_list/1e9,sigma_d,colors='black',levels=0)
    ax[1,0].contourf(kappa_list,Ap_list/1e9,np.ma.masked_where(np.logical_and(np.logical_not(stability_c),np.logical_not(stability_d)),sigma_d),hatches='/',levels=[0,500],alpha=0)
if np.min(sigma_c)<0:
    ax[1,0].contour(kappa_list,Ap_list/1e9,sigma_c,colors='black',levels=0,linestyles='dashed')

ax[1,0].scatter(kappa_diff_ref,Ap_ref/1e9,facecolors='white',edgecolors='black',marker='d',s=50,linewidths=1)

ax[1,0].set_xscale('log')
ax[1,0].set_xlabel(r'$\kappa_\mathrm{diff}$'+' [m/s]')


################################
#     Ap x kappa_conv plot     #
################################

kappa_list = np.logspace(-7,-2,Npoint)
kappa_position = 8

melt_d, melt_c, sigma_d, sigma_c, stability_d, stability_c, bistable = melt_map(Ap_list, kappa_list, kappa_position, 200, 0.1, regime="conv")

ax[1,1].pcolormesh(kappa_list/kappa_diff_ref,Ap_list/1e9,melt_c,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[1,1].pcolormesh(kappa_list/kappa_diff_ref,Ap_list/1e9,melt_d,cmap=lipari_cut,vmin=0,vmax=vmax,shading='gouraud') # Ap in km²
ax[1,1].pcolormesh(kappa_list/kappa_diff_ref,Ap_list/1e9,bistable,cmap='binary',vmin=0,vmax=1) # Ap in km²


if np.max(sigma_d)>0:
    ax[1,1].contour(kappa_list/kappa_diff_ref,Ap_list/1e9,sigma_d,colors='black',levels=0)
    ax[1,1].contourf(kappa_list/kappa_diff_ref,Ap_list/1e9,np.ma.masked_where(np.logical_and(np.logical_not(stability_c),np.logical_not(stability_d)),sigma_d),hatches='/',levels=[0,500],alpha=0)
if np.min(sigma_c)<0:
    ax[1,1].contour(kappa_list/kappa_diff_ref,Ap_list/1e9,sigma_c,colors='black',levels=0,linestyles='dashed')

ax[1,1].scatter(kappa_conv_ref/kappa_diff_ref,Ap_ref/1e9,facecolors='white',edgecolors='black',marker='d',s=50,linewidths=1)

ax[1,1].set_xscale('log')
ax[1,1].set_xlabel(r'$\kappa_\mathrm{conv}/\kappa_\mathrm{diff}$')
ax[1,0].set_ylabel(r'$\mathcal{A}_p$'+' '+r'$\mathrm{[\times 10^3 km^2]}$')


##########################
#     Ap x Lamb plot     #
##########################

Lamb_list = np.logspace(-2,0,Npoint)
Lamb_position = 10

monostable, bistable, sigma_fw, sigma_bw = melt_map_smooth(Ap_list, Lamb_list, Lamb_position, 400, 0.1, 0.01)

ax[1,2].pcolormesh(Lamb_list,Ap_list/1e9,monostable,cmap=lipari_cut,vmin=0,vmax=2,)
ax[1,2].pcolormesh(Lamb_list,Ap_list/1e9,bistable,cmap='binary',vmin=0,vmax=1)
ax[1,2].set_xscale('log')
ax[1,2].set_xlabel(r'$\Lambda$'+' '+r'$\mathrm{[kg/m^3]}$')



plt.tight_layout()
fig.subplots_adjust(right=0.8)
cbar_ax = fig.add_axes([0.85, 0.15, 0.03, 0.7])
fig.colorbar(plot1,cax=cbar_ax, label = 'Basal melt rate '+r'$m$'+' [m/30d]', extend='max')
plt.savefig('Figures/Figure4_PIG.png',dpi=300,format="png", bbox_inches = 'tight')
plt.show() 
```
